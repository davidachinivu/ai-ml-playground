Initialize Groq Client


```python
from google.colab import userdata
from groq import Groq

GROQ_API_KEY = userdata.get('GROQ_API_KEY')

client = Groq(api_key=GROQ_API_KEY)

print("Groq Client Initialized Successfully!")
```

Basic Text Generation with Groq


```python
prompt_text = "The future of AI in schools is"

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": prompt_text,
        }
    ],
    model="llama-3.3-70b-versatile", # Updated to a different supported Groq model
    max_tokens=500,
    temperature=0.7
)

print("--- Chat Completion Details ---")
print(f"Content: {chat_completion.choices[0].message.content}")


if hasattr(chat_completion.choices[0].message, 'reasoning') and chat_completion.choices[0].message.reasoning:
    print(f"Reasoning: {chat_completion.choices[0].message.reasoning}")



print("\nFull ChatCompletion Object (for debugging):")
print(chat_completion)
print("------------------------------")
```

#Agentic Workflow Demo

A simple/basic AI Agent has 3 steps:

Plan – Understand the task

Think – Generate a solution

Act – Show the result


```python
def ai_agent(task):
    print('PLAN: Understanding the task...')
    planning_prompt = f"Understand this task and explain what needs to be done: {task}"
    plan_response = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": planning_prompt,
            }
        ],
        model="llama-3.1-8b-instant", # Updated model
        max_tokens=1600, # Further increased max_tokens for planning
        temperature=0.7
    )
    print("--- Plan Response ---")
    print(f"Content: {plan_response.choices[0].message.content}")
    if hasattr(plan_response.choices[0].message, 'reasoning') and plan_response.choices[0].message.reasoning:
        print(f"Reasoning: {plan_response.choices[0].message.reasoning}")
    print("---------------------")

    print('\n THINK: Generating solution...')
    thinking_prompt = f"Solve this task step by step: {task}"
    thought_response = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": thinking_prompt,
            }
        ],
        model="llama-3.1-8b-instant", # Updated model
        max_tokens=2400, # Further increased max_tokens for thinking
        temperature=0.7
    )
    print("--- Thought Response ---")
    print(f"Content: {thought_response.choices[0].message.content}")
    if hasattr(thought_response.choices[0].message, 'reasoning') and thought_response.choices[0].message.reasoning:
        print(f"Reasoning: {thought_response.choices[0].message.reasoning}")
    print("------------------------")


    print('\n ACT: Final Answer')
    action_prompt = f"Give the final short answer for: {task}"
    action_response = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": action_prompt,
            }
        ],
        model="llama-3.1-8b-instant", # Updated model
        max_tokens=800, # Further increased max_tokens for acting
        temperature=0.7
    )
    print("--- Action Response ---")
    print(f"Content: {action_response.choices[0].message.content}")
    if hasattr(action_response.choices[0].message, 'reasoning') and action_response.choices[0].message.reasoning:
        print(f"Reasoning: {action_response.choices[0].message.reasoning}")
    print("-----------------------")
```

Testing the AI agent

```python
user_task = input("Enter a task for the AI Agent: ")
ai_agent(user_task)
```
